## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.


import pandas as pd
import numpy as np

# Load the FlyRank starter dataset
df = pd.read_csv("content_refresh_anonymized.csv")

# 1. Dataset size
print("Number of pages:", len(df))

# 2. Average click-through rate
print("Average CTR:", round(df["ctr"].mean(), 4))

# 3. Average position
print("Average position:", round(df["avg_position"].mean(), 2))

# 4. Correlation between position and CTR
correlation = df["avg_position"].corr(df["ctr"])
print("Correlation between average position and CTR:", round(correlation, 3))
```
