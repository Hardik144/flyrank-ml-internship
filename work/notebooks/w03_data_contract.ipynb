```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.


import pandas as pd

# 1. Check the total number of rows and columns
print("Total rows:", len(df))
print("Total columns:", len(df.columns))

# 2. Check for duplicate records
print("Duplicate rows:", df.duplicated().sum())

# 3. Check missing values
print("\nMissing values per column:")
display(df.isnull().sum().to_frame("Missing values"))

# 4. Check the date coverage without displaying individual records
date_columns = [
    col for col in df.columns
    if any(word in col.lower() for word in ["date", "day", "time"])
]

for col in date_columns:
    dates = pd.to_datetime(df[col], errors="coerce")
    if dates.notna().any():
        print(f"\nDate coverage for {col}:")
        print("Earliest date:", dates.min())
        print("Latest date:", dates.max())
        print("Valid date records:", dates.notna().sum())

# 5. Check whether the dataset contains missing or incomplete records
print("\nRows containing at least one missing value:",
      df.isnull().any(axis=1).sum())

print("\nData limits verification completed.")
```

## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

### Data limits

My analysis is limited to the data available in the FlyRank warehouse and the development time window I selected.

1. **Incomplete historical coverage:** The dataset may not contain enough historical observations to capture long-term trends or seasonal patterns.
2. **Unbalanced data:** Some content items may have more observations than others, so the available records may not represent every item equally.
3. **Availability limitations:** Some records may have missing or unavailable search performance data. These records cannot automatically be treated as zero performance.
4. **Limited generalization:** Results from my selected content and time window may not apply to all clients, content types, or future periods.
5. **Association is not causation:** Search performance patterns alone cannot prove that a particular content change caused a ranking or traffic change.

I will interpret the results as decision-support evidence rather than guaranteed predictions. I will also avoid using information that would not have been available at the time of the decision.
